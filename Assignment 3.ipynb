```python
def circle_area(radius):
    return math.pi * (radius ** 2)
```

```python
print(circle_area(10))
```

```python
def calculator(num1, num2, operation):
    if operation == 'a':
        return num1 + num2
    elif operation == 's':
        return num1 - num2
    elif operation == 'm':
        return num1 * num2
    elif operation == 'd':
        return num1 / num2
    else:
        return "Unknown operation"
```

```python
print(calculator(2, 5, 'd'))
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
        
    def area(self):
        return self.length * self.width
```

```python
r = Rectangle(5, 10)
print(r.area())
```

```python
class Shape:
    def __init__(self, name, length):
        self.name = name
        self.length = length
        
    def area(self):
        return 0

class Square(Shape):
    def __init__(self, name, length):
        super().__init__(name, length)
        
    def area(self):
        return f"The area is:\n{self.length ** 2}"
        
    def describe(self):
        return f"This is a: {self.name}"
```

```python
s = Square('square', 5)
print(s.area())
print(s.describe())
```
